```python

def total_rain(rain_mm: list[float]) -> float:
    """total rain over days in mm"""
    run_total = 0
    for mm in rain_mm:
        run_total += mm
    return run_total

def count_hot_days(temps: list[int]) -> int:
    """count hot days, 30 or more"""
    count = 0
    for t in temps:
        if t >= 30:
            count += 1
        return count

    def test_rain_and_heat():
        assert total_rain([2, 0, 5, 1]) == 8
        assert total_rain([]) == 0
        assert count_hot_days([28, 30, 32, 29]) == 2
        assert count_hot_days([]) == 0
```
